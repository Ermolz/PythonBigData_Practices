15. Порівняйте список та кортеж. Назвіть схожості та відмінності, випадки використання.

Списки (list) і кортежі (tuple) схожі тим, що є впорядкованими послідовностями, підтримують індексацію,
зрізи, перебір через for, можуть містити різні типи даних і використовують функції len(), count(),
index(). Відмінності: список змінюваний (mutable), а кортеж — ні (immutable). Кортеж швидший і займає
менше пам’яті, але не підтримує методи зміни (append(), remove(), sort() тощо). Список використовують,
коли дані змінюються, наприклад, список користувачів. Кортеж застосовують для незмінних даних, як-от
дні тижня або координати. Якщо потрібна гнучкість — вибирай list, якщо важлива швидкість і сталість
даних — tuple.

# VI. Множини (Set).

17. Створити множину без елементів. Після цого за допомогою методу додайте кілька різних елементів до множини. Чи множини є змінним типом даних?

```python
set1 = set()
set1.add(1)
set1.add(2)
set1.add(3)
print(set1)
```

Так, множини є змінним типом даних. Однак самі елементи множини повинні бути незмінними (наприклад, числа,
рядки, кортежі).
Для незмінної  множини, використовується frozenset, який не дозволяє змінювати вміст після створення
.*місце для текстової відповіді

18. Створити множину, маючи список my_list = [1, 1, 2, 67, 67, 8, 9]. Пояснити, чому "зникли" деякі елементи.

```python
my_list = [1, 1, 2, 67, 67, 8, 9]
my_set = set(my_list)
print(my_set)
```

У множині  зникають дублікати, тому що множина зберігає тільки унікальні значення. При конвертації списку
в множину **(set(my_list))** всі повторювані елементи автоматично видаляються.

19. Створіть дві множини. Продемонстуйте над ними операції: обʼєднання, різниці, пересічі та симетричної різниці. Використовуйте методи, що не змінюють множини, а створюють нові.

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

print("Union:", set1.union(set2))
print("Difference:", set1.difference(set2))
print("Intersection:", set1.intersection(set2))
print("Symmetric difference:", set1.symmetric_difference(set2))
```
