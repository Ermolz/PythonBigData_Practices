5. (9б). Створіть клас 3D-фігура без атрибутів, з методом get_volume для отримання об'єму фігури, що повертає 0 та __add __, який приймає self та other в якості аргументів, а повертає суму об'ємів 3d-фігур self та other.

```python
class Figure3D:
    def get_volume(self):
        return 0
    def __add__(self, other):
        if isinstance(other, Figure3D):
            return self.get_volume() + other.get_volume()
        raise TypeError("Можна додавати тільки об'єкти класу Figure3D")

figure1 = Figure3D()
figure2 = Figure3D()
print(figure1 + figure2)
```

6. (11б). Створіть 2 дочірніх класи від 3d-фігура: Циліндр та Куб, які мають атрибути, необхідні для розрахунку об'єму. Визначте метод get_volume в кожному з них так, щоби вони розраховували об'єм у залежності від формули для кожного типу фігури. Створіть обʼєкт класу Циліндр та обʼєкт класу Куб. Виконайте операцію суми за допомогою оператора + між ними, аби знайти сумарний об'єм.

```python
import math

class Figure3D:
    def get_volume(self):
        return 0

    def __add__(self, other):
        if isinstance(other, Figure3D):
            return self.get_volume() + other.get_volume()
        raise TypeError("Можна додавати тільки об'єкти класу Figure3D")

class Cylinder(Figure3D):
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height

    def get_volume(self):
        return math.pi * self.radius**2 * self.height

class Cube(Figure3D):
    def __init__(self, side):
        self.side = side

    def get_volume(self):
        return self.side**3

cylinder = Cylinder(3, 5)
cube = Cube(4)

total_volume = cylinder + cube
print(f"Сумарний об'єм: {total_volume:.2f}")
```

7. (3б). Продемонструйте різницю між isinstance та issubclass.

```python
class Animal:
    pass

class Dog(Animal):
    pass

dog = Dog()

# isinstance - перевіряє, чи є об'єкт екземпляром класу або його нащадка
print(isinstance(dog, Dog))
print(isinstance(dog, Animal))
print(isinstance(dog, int))

# issubclass - перевіряє, чи є один клас нащадком іншого
print(issubclass(Dog, Animal))
print(issubclass(Animal, Dog))
print(issubclass(Dog, object))
```

8. (13б). Створіть клас EWallet з приватними атрибутами balance та id. Реалізуйте методи поповнення та зняття коштів, забезпечивши належну інкапсуляцію. Підказка: використовуйте декоратори property та setter.

```python
import uuid

class EWallet:
    def __init__(self, balance=0):
        self.__balance = balance
        self.__id = str(uuid.uuid4())

    @property
    def balance(self):
        return self.__balance

    @property
    def id(self):
        return self.__id

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
        else:
            raise ValueError("Сума поповнення має бути більше 0")

    def withdraw(self, amount):
        if 0 < amount <= self.__balance:
            self.__balance -= amount
        else:
            raise ValueError("Недостатньо коштів або некоректна сума")

wallet = EWallet(100)
print(f"Wallet ID: {wallet.id}")
print(f"Balance: ${wallet.balance}")

wallet.deposit(50)
print(f"Balance after deposit: ${wallet.balance}")

wallet.withdraw(30)
print(f"Balance after withdrawal: ${wallet.balance}")
```

9. (11б). Створіть клас ShoppingCart, який містить список об'єктів типу Item. Реалізуйте методи для додавання айтемів, їх видалення та відображення списку айтемів.

```python
class Item:
    def __init__(self, name, price):
        self.name = name
        self.price = price

    def __str__(self):
        return f"{self.name}: ${self.price:.2f}"

class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, item):
        self.items.append(item)

    def remove_item(self, item_name):
        self.items = [item for item in self.items if item.name != item_name]

    def show_items(self):
        return "\n".join(str(item) for item in self.items) if self.items else "Cart is empty"

cart = ShoppingCart()
item1 = Item("Laptop", 1200)
item2 = Item("Phone", 800)

cart.add_item(item1)
cart.add_item(item2)

print(cart.show_items())

cart.remove_item("Laptop")
print(cart.show_items())
```

Ітерабельні об'єкти можна перебирати в циклі for, але самі по собі не зберігають стан ітерації, тоді як ітератор — це об'єкт, який реалізує методи \_\_iter__() та \_\_next__(), зберігає поточний стан ітерації та видає елементи послідовно (наприклад, iter([1, 2, 3])).

13. (11б). Створіть клас, що має визначений Iterator Protocol та при кожному виклику next повертає назву дня тижня, поки вони не вичерпаються. Зверніть увагу, що порядок має значення.

```python
class WeekdaysIterator:
    def __init__(self):
        self.days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.index < len(self.days):
            day = self.days[self.index]
            self.index += 1
            return day
        raise StopIteration

week_iterator = WeekdaysIterator()

for day in week_iterator:
    print(day)
```
